```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.get_dataset_names()
```

```python
df = sns.load_dataset('titanic')
df
```

```python
df.info()
```

```python
df['age'] = df['age'].fillna(0)
df.info()
```

```python
df['survived']=df['survived'].astype(np.int8)
df['pclass']=df['pclass'].astype(np.int8)
df['age']=df['age'].astype(np.int8)
df['sibsp']=df['sibsp'].astype(np.int8)
df['parch']=df['parch'].astype(np.int8)
df.info()
```

```python
df[['pclass', 'sex']].replace({
    'pclass': {1: 'First', 2: 'Second', 3: 'Third'}
}).value_counts()
```

```python
df['alive'].value_counts(normalize=True) * 100
```

```python
df['sex'].value_counts()
```

```python
df['pclass'].value_counts()
```

```python
df['who'].value_counts()
```

```python
df["alone"].value_counts()
```

```python
df["embark_town"].value_counts()
```

```python
df[['sex', 'alive']].replace({
    'alive': {
        'no': 'مرده',
        'yes': 'زنده'
    },
    'sex': {
        'male': 'مرد',
        'female': 'زن'
    }
}).value_counts(normalize=True)*100
```

```python
df[df['who'] == 'child'][['who', 'alive']].value_counts()
```

```python
df[df["alive"]=="yes"][df["sex"]=="female"][['alive','sex']].value_counts()
```

```python
df[(df["sex"]=="male")&(df["alive"]=="no")][["sex","alive"]].value_counts()
```

```python
# تعداد زنان و مردانی که در کلاس اول (pclass = 1) سفر می‌کردند را به دست بیاور.
df[df["pclass"]==1][["sex","pclass"]].value_counts()
```

```python
# تعداد کودکان زنده‌مانده را بر اساس جنسیت تفکیک کن.
df[(df["alive"]=="yes")&(df["who"]=="child")][["alive","who","sex"]].value_counts()
```

```python
# تعداد زنان و مردان فوت‌شده‌ای را پیدا کن که تنها سفر می‌کردند (alone = True).
df[(df["alone"]==True)&(df["alive"]=="no")][["sex"]].value_counts()
```

```python
# درصد زنده‌ماندگان را در هر کلاس بلیت (اول، دوم و سوم) محاسبه کن.
df[df["alive"]=="yes"][["pclass","alive"]].value_counts(normalize=True)*100
```

```python
# تعداد مسافران زنده‌مانده‌ای را به دست بیاور که از شهر Southampton سوار کشتی شده‌اند.
df[(df["embark_town"]=="Southampton")&(df["alive"]=="yes")][["embark_town","alive"]].value_counts()
```

```python
# تعداد مسافرانی را پیدا کن که سن آن‌ها ثبت نشده است.
df["age"].isna().sum()
```

```python
# تعداد مسافرانی را پیدا کن که سن آن‌ها ثبت شده است.
df["age"].notna().sum()
```

```python
# تعداد زنان و مردانی را که سن آن‌ها ثبت نشده است، جداگانه محاسبه کن.
df["sex"][df["age"].isna()].value_counts()
```

```python
# تعداد زنان و مردانی را که سن آن‌ها ثبت شده است، جداگانه محاسبه کن.
df["sex"][df["age"].notna()].value_counts()
```

```python
# تعداد مسافران بر اساس شهر مبدأ را فقط برای ردیف‌هایی محاسبه کن که شهر مبدأ آن‌ها خالی نیست
df['embark_town'][df["embark_town"].notna()].value_counts()
```

```python
# تعداد مسافرانی را پیدا کن که سن آن‌ها کمتر از ۱۸ سال است.
(df['age']<18).value_counts()
```

```python
#تعداد زنان و مردانی را که سن آن‌ها بیشتر از ۴۰ سال است، جداگانه محاسبه کن.
df[df['age']>40]['sex'].value_counts()
```

```python
#تعداد مسافرانی را پیدا کن که:
# زنده مانده‌اند.
#- سن آن‌ها بیشتر از ۳۰ سال است.
df[(df['alive']=='yes')&(df['age']>30)].shape[0]
```

```python
#3تعداد مسافران هر کلاس بلیت را فقط برای افرادی که زنده مانده‌اند، محاسبه کن.
df[df['alive']=='yes']['class'].value_counts()
```

```python
#تنها سفر نکرده‌اند (alone == False).
#زنده مانده‌اند (alive == yes).
df[(df['alive']=='yes')& (df['alone']==False)].shape[0]
```

```python
#تعداد زنان و مردانی را پیدا کن که:
#- در کلاس سوم بوده‌اند (pclass == 3)
#- فوت کرده‌اند (alive == 'no')
df[(df['alive']=='no')&(df['pclass']==3)]['sex'].value_counts()
```

```python
#تعداد مسافرانی را پیدا کن که مبلغ بلیت آن‌ها بیشتر از ۱۰۰ بوده است.
df[df['fare']>100].shape[0]
```

```python
#مسافرانی را پیدا کن که سن آن‌ها بین ۲۰ تا ۳۰ سال است (شامل خود ۲۰ و ۳۰).
df[(df['age']>=20)&(df['age']<=30)].shape[0]
```

```python
#درصد مسافران زن و مرد را محاسبه کن.
df['sex'].value_counts(normalize=True)*100
```
